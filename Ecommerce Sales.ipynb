```python
import numpy as np
import pandas as pd

df=pd.read_csv('Ecommerce project.csv')
df.head()
```

```python
#remove duplicate

df.drop_duplicates(inplace=True)
```

```python
df.duplicated().sum()
```

```python
#replace invalid value

df.replace(['N/A','NULL',' '],np.nan,inplace=True)
```

```python
#remove extra space
df=df.apply(lambda x: x.str.strip() if x.dtype=='str' else x)
```

```python
#standardize text format
df['Customer_Name']=df['Customer_Name'].str.title()

df['City']=df['City'].str.title()

df['State']=df['State'].str.title()
```

```python
#validate email addresses --keep only valid email

df=df[df['Email'].str.contains('@',na=False)]
```

```python
#Convert date column

df['Order_Date']=pd.to_datetime(df['Order_Date'],errors='coerce')
df['Delivery_Date']=pd.to_datetime(df['Delivery_Date'],errors='coerce')
```

```python
#convert numeric columns

cols=['Qty','Unit_Price','Discount']

for c in cols:
    df[c]=pd.to_numeric(df[c],errors='coerce')
```

```python
#Remove invalid quantity --Remove neative or zero quantity

df=df[df['Qty']>0]
```

```python
#fill missing value
df['Discount']=df['Discount'].fillna(0)
df['Phone']=df['Phone'].fillna('Unknown')
df['Delivery_Date']=df['Delivery_Date'].fillna(df['Order_Date'])
```

```python
#sales amount column
df['Sales']=df['Qty']*df['Unit_Price']
df
```

```python
#Net_amount column
df['Net_Amount']=df['Sales']-(df['Sales']*df['Discount']/100)
df
```

```python
#profit(20%) column
df['Profit']=df['Net_Amount']*0.20
df
```

```python
#order month,year,weekday
df['Month']=df['Order_Date'].dt.month_name()
df['Year']=df['Order_Date'].dt.year
df['Weekday']=df['Order_Date'].dt.day_name()
df
```

```python
df.columns=df.columns.str.lower()
df.columns=df.columns.str.replace(" ","_")
df.columns
```

```python
#Final data validation
df.info()
```

```python
df.isnull().sum() # we handle null value in power bi
```

```python
#export clean data
df.to_csv('Clean_Ecommerce.csv',index=False)
```
